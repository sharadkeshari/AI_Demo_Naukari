TF-IDF = Term Frequency - Inverse Document Frequency

TF (Term Frequency):
- Measures how often a word appears in a document.

IDF (Inverse Document Frequency):
- Measures how rare or unique a word is across all documents.

Formula:
TF-IDF = TF × IDF

Key Idea:
- High score → Word is frequent in a document but rare across documents.
- Low score → Word is common across many documents (e.g., "the", "is", "and").

Use Cases:
- Search engines
- Document retrieval
- Keyword extraction
- Text classification
- Resume and job matching

Interview One-Liner:
"TF-IDF is a statistical technique that identifies important words in a document by combining how frequently they appear in that document with how rare they are across a collection of documents."
